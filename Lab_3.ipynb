## Implement: Input and Output

### Subtask:
Generate Python code to get user input and print output to the console.


**Reasoning**:
The subtask requires generating Python code to get user input for name and age, convert the age to an integer, and then print a personalized greeting to the console. The code block will fulfill all these requirements.



````python
import sys

# 1. Prompt the user to enter their name
user_name = input("Enter your name: ")

# 2. Prompt the user to enter their age and convert it to an integer
# Using a try-except block to handle potential ValueError if input is not a valid integer.
try:
    user_age = int(input("Enter your age: "))
except ValueError:
    print("Invalid age entered. Please enter a number.", file=sys.stderr)
    user_age = 0 # Assign a default or handle as appropriate

# 3. Print a personalized greeting message
if user_age != 0: # Only print greeting if age was valid or handled
    print(f"Hello, {user_name}! You are {user_age} years old.")
else:
    print(f"Hello, {user_name}! We couldn't determine your age.")
````
